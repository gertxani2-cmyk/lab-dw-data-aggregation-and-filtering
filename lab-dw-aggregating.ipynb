```python
# your code goes here

import pandas as pd

# Load dataset
url = "https://raw.githubusercontent.com/data-bootcamp-v4/data/main/marketing_customer_analysis.csv"
df = pd.read_csv(url)

# Standardize column 
df.columns = df.columns.str.lower().str.replace(' ', '_')
```

```python
low_claims_responsive = df[
    (df['total_claim_amount'] < 1000) & 
    (df['response'] == 'Yes')
]
display(low_claims_responsive)
```

```python
# Aggregate profiles of converted customers
converted_analysis = (
    df[df['response'] == 'Yes']
    .groupby(['policy_type', 'gender'])[['monthly_premium_auto', 'customer_lifetime_value', 'total_claim_amount']]
    .mean()
)
display(converted_analysis)
```

```python
state_distribution = df.groupby('state').size()
large_markets = state_distribution[state_distribution > 500]

display(large_markets)
```

```python
clv_summary = (
    df.groupby(['education', 'gender'])['customer_lifetime_value']
    .agg(['max', 'min', 'median'])
)
display(clv_summary)
```

```python
# Convert index to a datetime format and isolate full Month name strings

df['effective_to_date'] = pd.to_datetime(df['effective_to_date'], format='%m/%d/%Y')
df['month'] = df['effective_to_date'].dt.strftime('%B')

# Generate the structural grid table layout
monthly_state_matrix = df.pivot_table(
    index='state', 
    columns='month', 
    values='customer', 
    aggfunc='count', 
    fill_value=0
)
display(monthly_state_matrix)
```

```python
# Isolate the index naming structures of the top 3 highest volume states
top_3_states = df['state'].value_counts().head(3).index.tolist()

# Filter and aggregate matching parameters
top_states_monthly = (
    df[df['state'].isin(top_3_states)]
    .groupby(['state', 'month'])
    .size()
    .reset_index(name='policies_sold')
)
display(top_states_monthly)
```

```python
#Bonus
# 1. Identify the top 3 states with the highest number of policies sold
top_3_states = df['state'].value_counts().head(3).index.tolist()

# 2. Filter the DataFrame to only include these top 3 states
df_top_3 = df[df['state'].isin(top_3_states)]

# 3. Group by state and month, then count the number of policies
policies_count = df_top_3.groupby(['state', 'month']).size().reset_index(name='policies_sold')

# 4. Reshape the data so months are columns and states are rows
monthly_state_table = policies_count.pivot(index='state', columns='month', values='policies_sold')
display(monthly_state_table)
```

```python
#bonus
# 1. Keep only the columns needed for marketing analysis
df_marketing = df[['sales_channel', 'response']]

# 2. Use melt to unpivot the data into a single channel column
melted_df = pd.melt(
    df_marketing, 
    id_vars=['response'], 
    value_vars=['sales_channel'], 
    value_name='marketing_channel'
)

# 3. Calculate the percentage of "Yes" responses for each channel
response_rate = pd.crosstab(melted_df['marketing_channel'], melted_df['response'], normalize='index') * 100
display(response_rate[['Yes']].round(2))
```
